```python
import seaborn as sns
import matplotlib.pyplot as plt

import pandas as pd

tips = pd.read_csv('tips.csv')
tips.head()

#tips.corr()

# corr_v1=tips[['total_bill', 'tip']]

# tips[['total_bill', 'tip', 'size']].sample(20)

fig = plt.figure(figsize = (6,3))
sns.heatmap(tips.corr(numeric_only=True), annot = True)

# sns.heatmap(tips[['total_bill', 'tip', 'size']].sample(20).corr(), annot = True)
```

```python
### countplot

#This is essentially the same as barplot except the estimator is explicitly counting the number of# occurrences. Which is why we only pass the x value:

tips.columns

tips['day'].value_counts()

#sns.countplot(data = tips, x = 'day' )

## boxplot and violinplot

# Can do entire dataframe with orient='h'
sns.boxplot(data = tips, palette = 'rainbow', orient = 'h')

sns.boxplot(data = tips, x = 'total_bill')

sns.boxplot(data = tips, x = 'day', y = "total_bill", hue = 'day', palette = 'pastel')

sns.boxplot(data = tips, x = "gender", y = "total_bill", palette = 'pastel', hue = 'gender')

sns.boxplot(x = "gender", y = "total_bill",
            hue = "day",
            data = tips,
            palette = "pastel", gap = 0.2)

plt.legend(loc = 'best')
```
