```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

df = pd.read_csv('tips.csv')

df.head(2)

df.describe([.05,.1,.2,.25])

df['total_bill'].quantile(.25)


total_bill = df[['total_bill']].copy()
total_bill

print(total_bill.min())
print(total_bill.max())

bins = range(0 , 60  , 5)

list(bins)

total_bill['bin / bucket'] = pd.cut( df['total_bill'] , bins , include_lowest = True , right = False  ) #  [0 , 5) [5 , 10)
# right = true = (0  , 5] , (5 , 10]

total_bill

total_bill['bin / bucket'].value_counts()  # occurences per bin

#print(total_bill['bin / bucket'].value_counts() )

df['total_bill']

bins = [0, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50, 55] # bins / buckets

plt.hist(df['total_bill'] , bins, histtype = 'bar', rwidth = 1 , alpha = 0.5 )

plt.xlabel('total bill intervals')
plt.xticks(bins)

plt.ylabel('freq / occurences')
plt.yticks(range(0,70 ,4))

plt.title('histogram')
plt.show()
```
