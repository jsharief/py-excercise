```python
import pandas as pd

sales_data = [120, 150, 130, 170, 160, 180, 140]

days_of_week = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']


df = pd.DataFrame(sales_data,days_of_week,columns=['Sales'])

df['average_sales'] =  df['Sales'] - df['Sales'].mean()


print(df)


```

```python
sales_data = [120, 150, 130, 170, 160, 180, 140]

days_of_week = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

sales_dict = { days_of_week[i]: sales_data[i] for i in range(len(sales_data)) }

avg_sales  = sum(sales_data) / len(sales_data)

sales_diff = {}

for day, sales in sales_dict.items():
    sales_diff = {
        'sales': sales,
        'sales_diff': sales - avg_sales
    }
    sales_dict[day] = sales_diff

print(sales_dict)

```
