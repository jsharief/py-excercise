```python
def greet(name,callback):
   callback(name)

def greet_in_english(name):
    print("Hello " + name)

def greetInspanish(name):
    print("Hola " + name)

def greetInArabic(name):
    print("Salam " + name)

def greetInFrench(name):
    print("Bonjour " + name)
def greetInHindi(name):
    print("Namaste " + name)

greet("John", greet_in_english)
greet("John",greetInspanish)
greet("John",greetInArabic)
greet("John",greetInFrench)
greet("John",greetInHindi)
```

```python
def swap_case(s):
  swapped = ""
  for x in s:
   if x.isupper():
        swapped += x.lower()
        s = s.replace(x, x.lower())
   else:
        swapped += x.upper()
        s = s.replace(x, x.upper())

  return s



print(swap_case("HackerRank.com presents Pythonist 2."))

```

```python
groceries = {
    "James": [10, 2, 5, 7, 9, 4, 6],
    "Tom": [1, 12, 3, 0, 1, 2, 5],
    "Sam": [9, 5, 4, 3, 2, 1, 3],
}

def most_purchases(groceries):
  x = {}
  for name,lst in groceries.items():
    tot = sum(lst)
    x[name] = tot

  print(max(x.items()))
  return max(x.items() , key=lambda item: item[1])

most_purchases(groceries)
```

```python
import numpy as np

# Read the dimensions (rows = X, columns = Y)
parts = input("Enter two integers (rows and columns) separated by space: ").split()
if len(parts) != 2:
    raise ValueError(f"Expected 2 values for rows and columns, got {len(parts)}")

X, Y = map(int, parts)

# Read the matrix rows
data = []
for i in range(X):
    row = list(map(int, input().split()))
    if len(row) != Y:
        raise ValueError(f"Expected {Y} columns in row {i+1}, got {len(row)}")
    data.append(row)

# Create the NumPy array
mat = np.array(data, dtype=int)

print("Matrix shape:", mat.shape)
print(mat)
```
